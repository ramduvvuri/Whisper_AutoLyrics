```python
import pandas as pd
import IPython.display as ipd
from pathlib import Path

df = pd.read_csv("../src/autolyrics/data/data/manifests/train.csv")

print(df.head())

# Listen to 5 random clips and read their text
for _, row in df.sample(5, random_state=0).iterrows():

    print(f"\n[{row.duration:.1f}s] {row.text}")

    audio_file = Path("..") / row.audio_path

    print("Audio:", audio_file)

    display(ipd.Audio(str(audio_file)))
```
